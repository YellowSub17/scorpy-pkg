```python
%load_ext autoreload
%autoreload 2
    
import numpy as np
import scorpy

import matplotlib.pyplot as plt



DET_Z = 110.0038/1e3

X,Y,Z,L,PHI =np.load(f'../mfx-cuso4-data/pix_map.npy')
PIX_SIZE = 75e-6
R = L*PIX_SIZE
THETA=0.5*np.arctan2(R,DET_Z)
Q = (4*np.pi/(0.794774))*np.sin(THETA)
D = 2*np.pi/Q




```

```python
import glob 

peak_files= glob.glob('../mfx-cuso4-data/*_peaks.npy')
c = scorpy.CorrelationVol(nq=256, npsi=360, qmin=1, qmax=8, cos_sample=False)
for peak_file in peak_files:
    print(f'\n\n{peak_file}')
    
    peak_frame, peak_x, peak_y, peak_i = np.load(peak_file).T

    #i_loc = np.where(peak_i<20e3)
    #peak_frame = peak_frame[i_loc]
    #peak_x = peak_x[i_loc]
    #peak_y = peak_y[i_loc]
    #peak_i = peak_i[i_loc]


    
    
    peak_frame = peak_frame.astype(int).flatten()
    peak_x = np.round(peak_x).flatten()
    peak_y = np.round(peak_y).flatten()
    
    
    peak_q = Q[peak_y.astype(int), peak_x.astype(int)]
    peak_phi = PHI[peak_y.astype(int), peak_x.astype(int)]

    
   
    for i_frame in range(np.max(peak_frame)+1):
        print(f'{i_frame}/{np.max(peak_frame)+1}', end='\r')
        frame_loc = np.where(peak_frame==i_frame)
        frame_q = peak_q[frame_loc]
        frame_phi = peak_phi[frame_loc]
        frame_i = peak_i[frame_loc]
    
        c.correlate_2D(frame_q, frame_phi, frame_i)

```

```python
c = scorpy.CorrelationVol(path='../cuso4-corr/cuso4_all_no_i_thresh.dbin')
#c.zmean_subtraction()
c.plot_q1q2()

c_blur = c.copy()
c_blur.convolve()
c_blur.plot_q1q2(vminmax=(0,1e11))
```

```python
cif = scorpy.CifData(a_mag=6.12, b_mag=10.7, c_mag=5.97, alpha=82.27, beta=107.43, gamma=102.67,spg='P-1')
cif.fill_from_vhkl(path='../mfx-cuso4-data/cuso4_sf_2.txt')
c_cif = scorpy.CorrelationVol(nq=256, npsi=360, qmin=1, qmax=8, cos_sample=False)
c_cif.fill_from_cif(cif)
_=c_cif.convolve()
```

```python
c_cif.plot_q1q2(vminmax=(0,1e6))
```

```python
fig, axs = plt.subplots(1,2, figsize=(16,8))
c_blur.plot_q1q2(vminmax=(0,5e11), fig=fig, axes=axs[0])
c_cif.plot_q1q2(vminmax=(0,1e7), fig=fig, axes=axs[1])

def plot_char_d(char_d, color):
    axs[1].plot(c_cif.psipts, char_d/(2*np.sin(c_cif.psipts/2)), color=color, linestyle='dashed')
    axs[1].set_ylim(c_cif.qmin, c_cif.qmax)
    axs[0].plot(c_cif.psipts, char_d/(2*np.sin(c_cif.psipts/2)), color=color, linestyle='dashed')
    axs[0].set_ylim(c_cif.qmin, c_cif.qmax)


plot_char_d(0.6, 'red')


plot_char_d(1.13, 'cyan')

plot_char_d(1.35, 'magenta')
#plot_char_d(2.6, 'magenta')
```
